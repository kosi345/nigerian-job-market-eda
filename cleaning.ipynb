Location mixed city+state pairs, "Rest of Nigeria (State)" tags, and non-geographic categories (Remote, Nationwide, Outside Nigeria, Unspecified). Standardized everything to a single consistent value per row (mostly state names) since a stakeholder to clarify design intent wasn't available — chose the simplest structure for EDA usability.

```python
import re

def clean_location(val):
    val = val.strip()
    if val.startswith('Rest of Nigeria ('):
        return val.split('(')[1].replace(')', '').strip()
    if val == 'Rest of Nigeria':
        return 'Unspecified'
    if '&' in val:
        return val.split('&')[1].replace('State', '').strip()
    return val

df['Location'] = df['Location'].apply(clean_location)
df['Location'].unique()
```

Salary combined currency, a text range, and special cases (Confidential, Commission Only, Less than/Above/More than) into one string. Split into currency, min_salary, max_salary (as floats, NaN where a bound is unknown), and salary_type to distinguish true ranges from non-numeric cases. Manually recoded one mislabeled row (NGN 150-300 → USD) that didn't match the NGN scale. Dropped the original Salary column after parsing.

```python
df.loc[df['Salary'] == 'NGN\n\n150 - 300', 'Salary'] = 'USD\n\n150 - 300'
```

```python
import re
import numpy as np

def parse_salary(val):
    currency, amount = val.split('\n\n')
    amount = amount.strip()
    
    if amount in ['Confidential', 'Commission Only']:
        return pd.Series([currency, np.nan, np.nan, amount])
    
    if amount.startswith('Less than'):
        num = float(re.sub(r'[^\d.]', '', amount))
        return pd.Series([currency, np.nan, num, 'Less than'])
    
    if amount.startswith('Above'):
        num = float(re.sub(r'[^\d.]', '', amount))
        return pd.Series([currency, num, np.nan, 'Above'])
    
    if amount.startswith('More than'):
        num = float(re.sub(r'[^\d.]', '', amount))
        return pd.Series([currency, num, np.nan, 'More than'])
    
    # standard range, e.g. "75,000 - 150,000"
    parts = amount.split('-')
    low = float(re.sub(r'[^\d.]', '', parts[0]))
    high = float(re.sub(r'[^\d.]', '', parts[1]))
    return pd.Series([currency, low, high, 'Range'])

df[['currency', 'min_salary', 'max_salary', 'salary_type']] = df['Salary'].apply(parse_salary)
```

```python
import re

def parse_details(s):
    quals = re.findall(r"Minimum Qualification\\n:\\n([^']+)", s)
    levels = re.findall(r"Experience Level\\n:\\n([^']+)", s)
    lengths = re.findall(r"Experience Length\\n:\\n([^']+)", s)
    
    qual = quals[0] if quals else 'Not Provided'
    level = levels[0] if levels else 'Not Provided'
    length = lengths[0] if lengths else 'Not Provided'
    
    return pd.Series([qual, level, length])

df[['min_qualification', 'experience_level', 'experience_length']] = df['details'].apply(parse_details)
```

date_posted used relative text ("2 weeks ago", "Today") with no confirmed scrape date available from the source repo, so an exact calendar date couldn't be reconstructed. Converted to days_ago (integer), treating weeks as 7 days and months as 30 days (approximate). Kept as a pure numeric column — unlike other cleaned columns, no "Not Provided" fallback was used here since every value in the source column matched a known pattern, and mixing text into a numeric column would break downstream math/sorting.

```python
df['date_posted'].unique()
```

```python
def parse_days_ago(val):
    val = val.strip()
    if val == 'Today':
        return 0
    if val == 'Yesterday':
        return 1
    match = re.match(r'(\d+)\s+(day|week|month)s?\s+ago', val)
    num, unit = int(match.group(1)), match.group(2)
    if unit == 'day':
        return num
    if unit == 'week':
        return num * 7
    return num * 30  # month

df['days_ago'] = df['date_posted'].apply(parse_days_ago)
df[['date_posted', 'days_ago']].drop_duplicates().sort_values('days_ago')
```
