"Filled missing hiring_firm_url and Industry with 'Not Provided' — these are non-numeric categorical gaps, not something to calculate around."

```python
df['hiring_firm_url'] = df['hiring_firm_url'].fillna('Not Provided')
df['Industry'] = df['Industry'].fillna('Not Provided')
```

```python
# 1. Fully identical rows (every column matches)
print("Fully identical duplicate rows:", df.duplicated().sum())

# 2. Duplicates based on job_url only (the real identity of a posting)
print("Duplicate job_url rows:", df.duplicated(subset='job_url').sum())

# 3. Look at one duplicated job_url and see ALL its copies side by side
dupe_url = df[df.duplicated(subset='job_url', keep=False)]['job_url'].iloc[0]
df[df['job_url'] == dupe_url]
```

Found 459 exact duplicate rows caused by a promoted/pinned listing reappearing on every scraped page (rows spaced ~19 apart, matching one page of results). Dropped with drop_duplicates() since every column matched — no information lost.

```python
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f"Removed {before - after} duplicate rows. New shape: {df.shape}")

df.duplicated().sum()   # should print 0
```
