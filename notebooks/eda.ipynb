## Most In-Demand Job Functions

Sales (315 postings) and Marketing & Communications (308) are the two most 
common job functions in this dataset, ahead of Engineering & Technology (232) 
and Software & Data (173).

A follow-up check showed Sales jobs are spread across 19 of 22 industries, 
rather than concentrated in one sector. This explains why Sales tops the list: 
it is a role nearly every type of business needs, not a role tied to one 
specific industry.

```python
df['job_function'].value_counts()
```

## Most In-Demand Industries

Excluding "Not Provided" (jobs with no industry listed), IT & Telecoms leads 
with 312 postings, followed by Hospitality & Hotel (267) and Banking, Finance 
& Insurance (255).

## Average Pay by Job Function

Engineering & Technology has the highest average pay (₦396,310), followed by 
Product & Project Management (₦377,451) and Management & Business Development 
(₦365,728).

Notably, Sales — the most in-demand job function — ranks only mid-table in pay 
(₦195,343), roughly half of Engineering & Technology. This shows that demand 
(number of postings) and pay do not always move together.

Averages are based only on postings in Naira with a full salary range stated. 
Categories with very few postings (e.g. Trades & Services, 3 jobs) should be 
treated as rough estimates, not reliable figures.

```python
ngn_range = df[(df['currency'] == 'NGN') & (df['salary_type'] == 'Range')].copy()
ngn_range['avg_salary'] = (ngn_range['min_salary'] + ngn_range['max_salary']) / 2

ngn_range.groupby('job_function')['avg_salary'].mean().sort_values(ascending=False)
```

```python
ngn_range.groupby('job_function')['avg_salary'].count().sort_values(ascending=False)
```

## Average Pay by Industry

Energy & Utilities has the highest average pay (₦449,537, based on 81 
postings), followed by IT & Telecoms (₦326,312) and Banking, Finance & 
Insurance (₦299,483).

Government and Enforcement & Security show the lowest postings in the clean 
salary data (1 and 3 jobs respectively) and their averages are not reliable.

```python
ngn_range.groupby('Industry')['avg_salary'].mean().sort_values(ascending=False)
```

```python
ngn_range.groupby('Industry')['avg_salary'].count().sort_values(ascending=False)
```

## Average Pay by Experience Level

Pay increases steadily and predictably with experience level:

- Internship & Graduate: ₦116,667
- No Experience: ₦119,792
- Entry level: ₦139,904
- Mid level: ₦208,355
- Senior level: ₦462,142
- Executive level: ₦711,328

Executive-level roles pay roughly 6 times more than Internship/Graduate roles, 
making experience level the strongest and most consistent salary driver found 
in this analysis.

```python
ngn_range.groupby('experience_level')['avg_salary'].agg(['mean', 'count']).sort_values('mean', ascending=False)
```

## Job Postings and Pay by Location

Lagos dominates job postings, accounting for 1,518 of 2,805 jobs (about 54%) — 
more than every other location combined. Abuja is a distant second (329), 
followed by Rivers State (116).

However, Lagos is not the highest-paying location. Despite having the most 
postings by far, Lagos ranks 10th in average pay (₦234,262). Rivers State has 
the highest reliable average pay (₦424,209, based on 79 postings), likely 
driven by its oil and gas industry presence — consistent with Energy & 
Utilities also being the top-paying industry overall.

This suggests Lagos offers the most job opportunities by volume, but not 
necessarily the best-paying ones.

```python
df["Location"].value_counts()
```

```python
ngn_range.groupby('Location')['avg_salary'].agg(['mean', 'count']).sort_values('mean', ascending=False)
```

## Average Pay by Minimum Qualification

Pay rises with required qualification level:

- High School (S.S.C.E): ₦120,833
- OND: ₦133,079
- Diploma: ₦138,133
- HND: ₦179,867
- Degree: ₦290,995
- MBA / MSc: ₦536,000

Degree is the most common requirement (887 clean postings) and pays notably 
more than HND or OND. Postgraduate qualifications (MBA/MSc, MPhil/PhD, MBBS) 
have very few postings in this dataset and should be read as indicative only, 
not conclusive.

```python
df['min_qualification'].value_counts()
```

```python
ngn_range.groupby('min_qualification')['avg_salary'].agg(['mean', 'count']).sort_values('mean', ascending=False)
```
